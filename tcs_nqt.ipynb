```python
def min_calories(calories, K):
    N = len(calories)

    # If impossible to pick K non-consecutive items
    if K > (N + 1) // 2:
        return -1

    # Create DP table
    dp = [[float('inf')] * (K + 1) for _ in range(N + 1)]

    # 0 items -> 0 calories
    for i in range(N + 1):
        dp[i][0] = 0

    for i in range(1, N + 1):
        for j in range(1, K + 1):

            # Option 1: Don't pick this item
            dp[i][j] = dp[i-1][j]

            # Option 2: Pick this item (skip previous)
            if i >= 2:
                dp[i][j] = min(dp[i][j], dp[i-2][j-1] + calories[i-1])
            elif j == 1:
                dp[i][j] = min(dp[i][j], calories[i-1])

    result = dp[N][K]

    return result if result != float('inf') else -1


# Example
calories = [100, 200, 50, 400, 150]
K = 2

print(min_calories(calories, K))
```

```python
# A team of researchers is monitoring oxygen levels in 3 different chambers across 3 rounds.
#  The average oxygen level of each chamber across rounds is calculated. 
#  The chamber with the highest average is declared the most fit.
#   If there's a tie, all chambers with the highest average are fit.
#    Oxygen levels must be between [1, 100]. 
# If any level is outside this range, the input is INVALID.
```

```python
a=list(map(int,input().split()))
b=list(map(int,input().split()))
c=list(map(int,input().split()))

all_values=a+b+c

for i in all_values:
    if i<1 or i>100:
        print("they are invalid")
        exit()

avg1=sum(a)/3        
avg2=sum(b)/3   
avg3=sum(c)/3   

highest=max(avg1,avg2,avg3)

if avg1 == highest:
    print("Chamber 1")

if avg2 == highest:
    print("Chamber 2")

if avg3 == highest:
    print("Chamber 3")
```
